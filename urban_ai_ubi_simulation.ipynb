# 城市空间演进模拟：AI-Agent 与 UBI

本 Notebook 展示一个可复现的城市空间演化仿真流程：
- 在网格城市中模拟家庭迁移与聚集；
- 设置 AI-Agent 覆盖率与 UBI 水平；
- 对比政策情景下的可负担性、空间不平等与就业机会；
- 给出模拟验证结果与可视化。

## 1. 系统说明（System Design）

### 1.1 状态变量
- 空间单元：`grid_size x grid_size` 二维网格。
- 家庭代理：收入、位置。
- 区位属性：租金面、可达性面、就业机会面。

### 1.2 行为机制
每个家庭在每个时间步比较全局区位效用，并迁移到效用最大单元。

### 1.3 政策冲击
- AI-Agent：提升就业匹配效率、降低搜索与通勤摩擦。
- UBI：提高收入底线，提升租金可负担性。

### 1.4 验证思路
- AI 提高平均就业机会；
- UBI 提高可负担性；
- AI+UBI 一般缓解空间不平等。
